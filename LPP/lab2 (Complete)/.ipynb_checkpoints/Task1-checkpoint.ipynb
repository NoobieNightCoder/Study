## Подключение библиотек

```python
import ship_table as st
```

## Считывание файла

```python
ERROR_STATUS = False
try:
    df = st.read_data("Cleaned_ships_data_broken.csv")
    st.file_check(df)
except Exception as e:
    print("Ошибка:" + str(e))
    ERROR_STATUS = True
```

## График зависимости средней максимальной загрузки кораблей от длины корабля

```python
try:
    if not ERROR_STATUS:
        fig = st.plot_average_dwt_by_length(df)
        fig.show()
except Exception as e:
    print("Ошибка:" + str(e))
    ERROR_STATUS = True
```

## Распределение количества кораблей по их типу 

```python
try:
    if not ERROR_STATUS:
        fig = st.plot_ship_types(df)
        fig.show()
except Exception as e:
    print("Ошибка:" + str(e))
    ERROR_STATUS = True
```

## Гистограмма распределения количества кораблей по годам постройки 

```python
try:
    if not ERROR_STATUS:
        fig = st.plot_ships_by_year(df)
        fig.show()
except Exception as e:
    print("Ошибка:" + str(e))
    ERROR_STATUS = True
```

## 2-D Гистограмма распределения количества кораблей по годам постройки и максимальному весу груза

```python
try:
    if not ERROR_STATUS:
        fig = st.plot_year_dwt_histogram(df)
        fig.show()
except Exception as e:
    print("Ошибка:" + str(e))
    ERROR_STATUS = True
```
