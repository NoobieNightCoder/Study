## Подключение библиотек

```python
import file_reader as fr
import iris_table as it
import matplotlib.pyplot as plot
```

## Считывание файла

```python
ERROR_STATUS = False
try:
    df = fr.read_file("IRIS.csv")
    it.file_check(df)
except Exception as e:
    print("Ошибка:" + str(e))
    ERROR_STATUS = True
```

## Парная диаграмма характеристик ирисов

```python
try:
    if not ERROR_STATUS:
        fig = it.plot_pairplot(df)
        fig.show()
except Exception as e:
    print("Ошибка:" + str(e))
    ERROR_STATUS = True
```

## График распределения характеристик ирисов по видам

```python
try:
    if not ERROR_STATUS:
        fig = it.plot_violinplot(df)
        fig.show()
except Exception as e:
    print("Ошибка:" + str(e))
    ERROR_STATUS = True
```
