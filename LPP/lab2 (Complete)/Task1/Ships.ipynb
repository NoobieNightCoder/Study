## Считывание файла

```python
ERROR_STATUS = False
try:
    df = fr.read_file("Cleaned_ships_data_test.csv")
    st.file_check(df)
except Exception as e:
    print("Ошибка:" + str(e))
    ERROR_STATUS = True
```

```python
try:
    if not ERROR_STATUS:
        fig = st.plot_average_dwt_by_length(df)
        fig.show()
except Exception as e:
    print("Ошибка:" + str(e))
    ERROR_STATUS = True
```
