## Считывание файла

```python
ERROR_STATUS = False
try:
    df = fr.read_file("pubg_stats.csv")
    pt.file_check(df)
except Exception as e:
    print("Ошибка:" + str(e))
    ERROR_STATUS = True
```

## График зависимости среднего числа убийств от количества сыгранных матчей

```python
try:
    if not ERROR_STATUS:
        fig = pt.plot_matches_played_kills(df)
        fig.show()
except Exception as e:
    print("Ошибка:" + str(e))
    ERROR_STATUS = True
```

### Данный график показывает, как количество сыгранных матчей связано с результативностью игроков. В данном случае более опытные игроки в среднем показывают более высокую результативность, чем менее опытные.

## График распределения количества рангов по их типу

```python
try:
    if not ERROR_STATUS:
        fig = pt.plot_rank(df)
        fig.show()
except Exception as e:
    print("Ошибка:" + str(e))
    ERROR_STATUS = True
```

### На данном графике можно увидеть соотношение игроков по рангам. В данном случае преобладает количество игроков с рангом Platinum, что не является самым высоким рангом в игре. Однако процент игроков ранга Diamond (что выше Platinum) тоже не маленький.

## Гистограмма распределения игроков по времени выживания

```python
try:
    if not ERROR_STATUS:
        fig = pt.plot_time_survived(df)
        fig.show()
except Exception as e:
    print("Ошибка:" + str(e))
    ERROR_STATUS = True
```

### Данный график показывает распределение длительности выживания игроков. На данном графике можно заметить, что более 20 игроков из таблицы в среднем за матч живут от 1.9 (114с) до 2 минут. Все игроки в среднем живут от 1.55 (93с) до 2.05 минут (123с)
