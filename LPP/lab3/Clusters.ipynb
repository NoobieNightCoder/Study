## KMeans Clustering

```python
if not ERROR_STATUS:
    main.kmeans_clustering(X, clusters_count)
```

```python
if not ERROR_STATUS:
    main.agl_clustering(X, clusters_count)
```

## DBScan

```python
if not ERROR_STATUS:
    main.dbscan_clustering(X)
```
